**Linear** **regression**

```python
class l_r():



  def __init__(self,learning_rate,no_of_itration):
    # declearing no of iteration and learning rate

      self.learning_rate=learning_rate
      self.no_of_itration=no_of_itration



  # fit funtion to train model with dataset or on dataset

  def fit(self,x,y):
      # here m = no of rows of dataset and n = no of column of dataset which we use futher.

      self.m,self.n=x.shape

  # initiating weight and bias


      self.w=np.zeros(self.n)
      self.b=0
      self.x=x
      self.y=y



      #Implementing gradiant decent

      for i in range(self.no_of_itration):

          self.update_weight()



  def update_weight(self):
        y_predtion = self.predict(self.x)
        #calculate gradients
        dw= -(2*(self.x.T).dot(self.y-y_predtion))/self.m
        db = -2*np.sum(self.y-y_predtion)/self.m

        #updating the weight
        self.w=self.w-self.learning_rate*dw
        self.b=self.b-self.learning_rate*db


  def predict(self,x):
    return x.dot(self.w) + self.b
```

```python
class logi_r :

# declearing no of iteration and learning rate
    def __init__(self, no_of_iteration,l_rate):
      self.l_rate = l_rate
      self.no_of_iteration = no_of_iteration
# fit funtion to train model with dataset or on dataset
    def fit(self,x,y):
      # here m = no of rows of dataset and n = no of column of dataset which we use futher.
      self.m ,self.n = x.shape
# initiating weight and bias
      self.w = np.zeros(self.n)
      self.b = 0
      self.x = x
      self.y = y

      for i in range(self.no_of_iteration):
        self.up_wt()


    def up_wt(self):
      # y_hat formula (sigmoid funtion )


       y_hat = 1/(1+np.exp(-(self.x.dot(self.w) + self.b))) # (self.x.dot(self.w) + self.b) = w.x +b

       dw = (1/self.m)*np.dot(self.x.T ,(y_hat - self. y))
       db = (1/self.m)*np.sum(y_hat - self. y)

# updating weight and bias using gradiant descent

       self.w = self.w - self.l_rate*dw
       self.b = self.b - self.l_rate*db




    def pred(self,x):
      y_pred = 1/(1+np.exp(-(x.dot(self.w) + self.b)))
      y_pre = np.where(y_pred>0.5,1,0)
      return y_pre
```

```python
import numpy as np
```

```python
class svm():

 def __init__(self,no_of_itt,ln_rate,lamda):# no of interation , learning rate and lamda

  self.no_of_itt = no_of_itt
  self.ln_rate= ln_rate
  self.lamda=lamda

 def fit(self,x,y):
  self.m , self.n = x.shape # m = no. rows and data point , n = no of columns and feature
  self.w=np.zeros(self.n)
  self.b= 0
  self.x = x
  self.y = y
  # implementing gradiant descent algo for optimization

  for i in range(self.no_of_itt):
    self.updateW()




 def updateW(self): # updating weight and bias in this funtion

  # label encoding of target variable
  y_label = np.where(self.y <=0,-1,1)

  for index , x_i in enumerate(self.x):
      ''' here , index which store the index of-- X and x_i-- which store value of x
      and these will done by help of -- enumerate(self.x) where enumerate is used as counter of variable X '''
      condition = y_label[index]*(np.dot(x_i,self.w)-self.b) >= 1
      if (condition == True):

         dw = 2*self.lamda*self.w
         db = 0

      else:
            dw = 2*self.lamda*self.w - np.dot(x_i,y_label[index])
            db = y_label[index]

      self.w = self.w - (self.ln_rate* dw)
      self.b = self.b- (self.ln_rate*db)






 def predict(self,x):
  output = np.dot(x,self.w)-self.b
  pred_label= np.sign(output)
  y_hat = np.where(pred_label <= -1,0,1)

  return y_hat
```
