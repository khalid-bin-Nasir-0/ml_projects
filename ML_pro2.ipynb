# Spam mail pred

importing dependencies

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

Data collection

```python
mail_data = pd.read_csv("/content/drive/MyDrive/ML dataset/mail_data.csv")
```

```python
mail_data.head()
```

```python
mail_data.isnull().sum()
```

```python
mail_data.shape
```

label encoding -- category spam - 0, ham - 1

```python
mail_data.loc[mail_data['Category'] == 'spam','Category',]=0
mail_data.loc[mail_data['Category'] == 'ham','Category',]=1
```

```python
mail_data.head()
```

```python
x = mail_data['Message']
```

```python
y = mail_data['Category']
```

```python
print(y)
```

spliting data into training and test daat

```python
x_train , x_test , y_train,y_test = train_test_split(x,y,test_size=0.2 , random_state=2)
```

```python
# transform the text data to feature vectors that can be used as input to logistic regression model
feature_extraction=TfidfVectorizer(min_df=1,stop_words='english',lowercase=True)

x_train = feature_extraction.fit_transform(x_train)
x_test= feature_extraction.transform(x_test)

print(x_train)
```

```python
# convert y_train and y_test value as integers

y_train = y_train.astype('int')
y_test = y_test.astype('int')


print(y_train)
```

training the model

```python
model = LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
train_pred = model.predict(x_train)
```

```python
accuracy = accuracy_score(y_train,train_pred)
```

```python
print(accuracy)
```

```python
test_pred = model.predict(x_test)
accuracy = accuracy_score(y_test,test_pred)
print(accuracy)
```
